### Gerando Embeddings

```python
from llama_index.embeddings.huggingface import HuggingFaceEmbedding
```

```python
class ChromaEmbeddingWrapper:
    def __init__(self, model_name):
        self.model = HuggingFaceEmbedding(model_name=model_name)
        
    def __call__(self, input):
        return self.model.embed(input)
```

```python
embed_model_chroma = ChromaEmbeddingWrapper(model_name="intfloat/multilingual-e5-large")
```

```python
import chromadb

db = chromadb.PersistentClient(path="./chroma.db")
chroma_client = db
```

```python
collection_name = "documentos_llm"
try:
    chroma_collection = chroma_client.get_or_create_collection(
        name=collection_name,
        embedding_function=embed_model_chroma
    )
except Exception as e:
    print(f"Erro ao carregar ou criar embeddings {e}")
```

### Salvando Embeddings no BD

```python
from llama_index.vector_stores.chroma import ChromaVectorStore
from llama_index.core import StorageContext
```

```python
vector_store = ChromaVectorStore(chroma_collection)
storage_context = StorageContext.from_defaults(vector_store=vector_store)
```

```python
embed_model = HuggingFaceEmbedding(model_name="intfloat/multilingual-e5-large")
```

```python
from llama_index.core import VectorStoreIndex

index = VectorStoreIndex(nodes, storage_context=storage_context, embed_model=embed_model)
```

```python
from llama_index.core import load_index_from_storage

index = load_index_from_storage(storage_context=storage_context, embed_model=embed_model)
```

### Interface Gráfica com Gradio

```python
import gradio as gr
```

```python
def chat_bot(message, chat_history):
    response = chat_engine.chat(message).response
    
    if chat_history is None:
        chat_history = []
        
    chat_history.append({"role":"user", "content": message})
    chat_history.append({"role":"assistant", "content": response})
    
    return "", chat_history

def reset_chat():
    chat_engine.reset()
    return []
```

```python
with gr.Blocks() as demo:
    gr.Markdown("# Chatbot de LLM")
    chatbot = gr.Chatbot(type="messages")
    msg = gr.Textbox(label="Digite a sua mensagem")
    resetar = gr.Button("Limpar")
    msg.submit(chat_bot, [msg, chatbot], [msg, chatbot])
    resetar.click(reset_chat, None, chatbot, queue=False)

demo.launch(debug=True)
```
