```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage

# Carregar variáveis de ambiente
load_dotenv()
guia = ChatOpenAI(api_key = os.getenv("OPENAI_API_KEY"),
                  model="gpt-5-mini", temperature=0.7)
```

### Exemplo 1 - Pergunta Simples

```python
resposta = guia.invoke("Quais são os pontos turísticos imperdíveis em Lisboa?")
print("🌍 Resposta simples:\n", resposta.content)
```

### Exemplo 2 - Conversa Estruturada

```python
mensagens = [
    SystemMessage(content="Você é um guia de viagens divertido e empolgado."),
    HumanMessage(content="Me recomende um prato típico de Tóquio e onde experimentá-lo.")
]

resposta = guia.invoke(mensagens)
print("🍣 Recomendação estruturada:\n", resposta.content)
```

### Exemplo 3 - Streaming
- Contando uma história de viagem em tempo real

```python
prompt = "Crie uma mini-história sobre uma aventura inesperada em Roma."
print("📖 História em streaming:\n")
for chunk in guia.stream(prompt):
    print(chunk.content, end="", flush=True)
```

### Mini Caso de Uso Prático
- Criando um "Guia de Viagens Virtual" que pode responder em diferentes estilos

```python
def guia_viagens(pergunta: str, estilo: str = "guia"):
    """
    Guia de Viagens Virtual que responde em estilos diferentes.
    Estilos: 'guia', 'poeta', 'historiador'
    """
    estilos = {
        "guia": "Você é um guia turístico simpático que dá dicas práticas de viagem.",
        "poeta": "Você é um poeta viajante que descreve os lugares de forma lírica.",
        "historiador": "Você é um historiador que explica curiosidades e fatos marcantes do destino."
    }
    mensagens = [
        SystemMessage(content=estilos.get(estilo, estilos["guia"])),
        HumanMessage(content=pergunta)
    ]
    return guia.invoke(mensagens).content
```

```python
print("🧭 Guia prático:\n", guia_viagens("O que visitar em Buenos Aires em 3 dias?", estilo="guia"))
```

```python
print("\n🎭 Poético:\n", guia_viagens("O que visitar em Buenos Aires em 3 dias?", estilo="poeta"))
```

```python
print("\n📜 Histórico:\n", guia_viagens("O que visitar em Buenos Aires em 3 dias?", estilo="historiador"))
```
