```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate, FewShotPromptTemplate

# Carregar variáveis de ambiente
load_dotenv()
chat = ChatOpenAI(api_key = os.getenv("OPENAI_API_KEY"),
                  model="gpt-5-mini", temperature=0.7)
```

### 1) Instruções claras (Clear Instructions) com Few-Shot
- Damos 2 exemplos de “explicar para uma criança” para fixar o estilo.

```python
exemplos_clareza = [
    {
        "conceito": "API",
        "explicacao": "Uma API é como um garçom: você pede algo ao sistema e ele traz o que você pediu."
    },
    {
        "conceito": "Nuvem",
        "explicacao": "A nuvem é como um armário online onde você guarda coisas e pode acessar de qualquer lugar."
    },
]

exemplo_template_clareza = PromptTemplate.from_template(
    "Pergunta: Explique {conceito} para uma criança de 10 anos.\nResposta: {explicacao}\n"
)

few_shot_clareza = FewShotPromptTemplate(
    examples=exemplos_clareza,
    example_prompt=exemplo_template_clareza,
    suffix="Pergunta: Explique {conceito} para uma criança de 10 anos.\nResposta:",
    input_variables=["conceito"],
)
```

```python
mensagem = few_shot_clareza.format(conceito="Inteligência Artificial")
resp = chat.invoke(mensagem)
print("👶 Instruções claras (few-shot):\n", resp.content)
```

### 2. Papéis e personas (Role Prompting)
- Definindo o "tom" e a identidade do modelo.


```python
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um professor de história engraçado que sempre conta piadas."),
    ("human", "Explique a queda do Império Romano em poucas frases.")
])

resposta = chat.invoke(chat_prompt.format_messages())
print("🎭 Persona aplicada:\n", resposta.content)
```

### 3. Constrained Output (formato estruturado)
- Pedindo a saída em formato específico (lista numerada).

```python
prompt = PromptTemplate.from_template(
    "Liste 3 dicas práticas para melhorar a {habilidade}, em formato de lista numerada."
)

mensagem = prompt.format(habilidade="memória")
resposta = chat.invoke(mensagem)
print("✅ Saída estruturada:\n", resposta.content)
```

### 4. Pensar passo a passo (Chain-of-Thought simplificado)
- Pedindo raciocínio explícito em etapas.

```python
prompt = PromptTemplate.from_template(
    "Explique passo a passo como funciona o processo de {processo}, "
    "e depois dê uma resposta final resumida em uma frase."
)

mensagem = prompt.format(processo="fotossíntese")
resposta = chat.invoke(mensagem)
print("🪴 Raciocínio passo a passo:\n", resposta.content)
```

### 5. Estilo de resposta (Style Prompting)
- Controlando o estilo/tom da saída.

```python
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um poeta que escreve respostas em forma de haikai."),
    ("human", "Descreva a importância da Lua para a Terra.")
])

resposta = chat.invoke(chat_prompt.format_messages())
print("🌙 Estilo aplicado (haikai):\n", resposta.content)
```
