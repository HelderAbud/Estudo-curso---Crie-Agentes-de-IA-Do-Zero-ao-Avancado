```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate
from langchain.memory import ConversationBufferMemory
from langchain.chains import LLMChain, SequentialChain, SimpleSequentialChain, ConversationChain


# Carregar variáveis de ambiente
load_dotenv()
chat = ChatOpenAI(api_key = os.getenv("OPENAI_API_KEY"),
                  model="gpt-5-mini", temperature=0.7)
```

### ConversationChain

```python
produto = "AuraSync, um assistente de IA para produtividade"
# Cria a memória
memory = ConversationBufferMemory(return_messages=True)

# Define o prompt (com o template padrão)
conversation = ConversationChain(
    llm=chat,
    memory=memory,
    verbose=True
)
```

```python
print(conversation.predict(input="Olá, vamos conversar sobre um produto de IA que ajuda na produtividade."))
```

```python
print(conversation.predict(input="Ele é chamado AuraSync. Qual tipo de pessoa se beneficiaria mais dele?"))
```

```python
print(conversation.predict(input="Qual é o nome do produto que estou te perguntando?"))
```

```python
print(memory.load_memory_variables({}))
```

### LLMChain
- A **`LLMChain`** é a cadeia mais básica, que conecta um prompt a um modelo de linguagem. É a base para a maioria das operações.

```python
prompt_slogan = PromptTemplate.from_template(
    """
    Crie um slogan criativo e memorável para o produto {produto}.
    """
)

# Cria a LLMChain
chain_slogan = LLMChain(llm=chat, prompt=prompt_slogan, verbose=True)
```

```python
slogan = chain_slogan.invoke({"produto": produto})
print("\n--- Slogan Gerado ---")
print(slogan['text'])
```

### SequentialChain
- A **`SequentialChain`** oferece mais controle, permitindo que você especifique as variáveis de entrada e saída em cada etapa.



```python
chain_nome = LLMChain(
    llm=chat,
    prompt=PromptTemplate.from_template("Gere 3 nomes para uma campanha de marketing do produto {produto}."),
    output_key="nome_campanha"
)

# Segunda etapa: gerar a descrição
chain_descricao = LLMChain(
    llm=chat,
    prompt=PromptTemplate.from_template("Para o produto {produto}, com a campanha {nome_campanha}, escreva uma descrição de até 50 palavras."),
    output_key="descricao_campanha"
)

# Terceira etapa: definir o público-alvo
chain_publico = LLMChain(
    llm=chat,
    prompt=PromptTemplate.from_template("Com base na descrição {descricao_campanha}, defina o público-alvo em uma frase."),
    output_key="publico_alvo"
)
```

```python
# Cria a cadeia sequencial completa
chain_sequencial_completa = SequentialChain(
    chains=[chain_nome, chain_descricao, chain_publico],
    input_variables=["produto"],
    output_variables=["nome_campanha", "descricao_campanha", "publico_alvo"],
    verbose=True
)
```

```python
resposta = chain_sequencial_completa.invoke({"produto": produto})
```

```python
print(f"Nome da Campanha: {resposta['nome_campanha']}")
```

```python
print(f"Descrição: {resposta['descricao_campanha']}")
```

```python
print(f"Público-alvo: {resposta['publico_alvo']}")
```
