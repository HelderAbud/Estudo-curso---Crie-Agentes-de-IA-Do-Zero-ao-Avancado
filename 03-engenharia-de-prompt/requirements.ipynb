```python
# Criando uma API client com verificação de chave
import os
from anthropic import Anthropic
api_key = os.getenv('ANTHROPIC_API_KEY')
if not api_key:
    raise RuntimeError('ANTHROPIC_API_KEY not found. Execute the first cell or add .env')
client = Anthropic(api_key=api_key)
model = "claude-3-5-sonnet-20241022"
print('Client criado')
```

```python
import sys
print(sys.executable)
```

```python
# Make a Request to the API (wrapped to show errors clearly)
try:
    response = client.messages.create(
        model=model,
        max_tokens=100,
        messages=[{"role": "user", "content": "What is the capital of Brazil?"}],
    )
    # normalize message variable for later cells
    try:
        message = response.content[0].text
    except Exception:
        message = str(response)
    print('Request succeeded. message set.')
except Exception as e:
    print('Request failed:', type(e).__name__, e)
    import traceback
    traceback.print_exc()
```

```python
# Mostrar a variável 'message' se existir
try:
    print('message =', message)
except NameError:
    print("message não definido. Execute a célula de request primeiro.")
```
